```python
import cv2
import os

# ====== 在这里修改视频路径 ======
VIDEO_PATH = "/home/oy/桌面/rctest/captured_video_rs1/realsense_record1.mp4"

# ====== 图片保存目录 ======
SAVE_DIR = "frames1"

# ====== 每隔多少帧保存一张 ======
INTERVAL = 3

# ====== 图片名前缀（可修改） ======
PREFIX = "oyhd6071_"


def main():
    os.makedirs(SAVE_DIR, exist_ok=True)

    cap = cv2.VideoCapture(VIDEO_PATH)

    if not cap.isOpened():
        print("无法打开视频:", VIDEO_PATH)
        return

    frame_id = 0
    save_id = 1   # 从1开始编号

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        if frame_id % INTERVAL == 0:
            filename = f"{PREFIX}{save_id:08d}.jpg"
            save_path = os.path.join(SAVE_DIR, filename)

            cv2.imwrite(save_path, frame)

            print("保存:", save_path)

            save_id += 1

        frame_id += 1

    cap.release()

    print("抽帧完成")
    print("总帧数:", frame_id)
    print("保存图片数:", save_id - 1)


if __name__ == "__main__":
    main()
```
